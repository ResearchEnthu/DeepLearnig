```python
from PIL import Image
import torchvision.transforms as transforms

prepocess = transforms.Compose([transforms.Resize(256),
                                 transforms.CenterCrop(224),
                                 transforms.ToTensor(),
                                 transforms.Normalize(mean=[0.485, 0.456, 0.406],
                                                      std=[0.229, 0.224, 0.225])])
```

```python
img = Image.open(r"C:\Users\alokk\OneDrive\Pictures\Screenshots 1\Screenshot 2026-06-21 153140.png").convert("RGB") 
img 
```

```python
img_t = prepocess(img)
batch_t = torch.unsqueeze(img_t, 0)
```

```python
batch_out = netG(batch_t)
```

```python
out_t = batch_out.detach().squeeze()

if out_t.ndim == 1:
	# ResNet output: class logits (not an image)
	probs = torch.nn.functional.softmax(out_t, dim=0)
	top_prob, top_idx = torch.topk(probs, 5)
	list(zip(top_idx.tolist(), top_prob.tolist()))
else:
	# Image-like tensor case
	out_t = ((out_t + 1.0) / 2.0).clamp(0, 1).cpu()
	out_img = transforms.ToPILImage()(out_t)
	out_img
	
```
